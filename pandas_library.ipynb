```python
# Pandas ==> jadvallar bilan ishlash, DataFrame va Series, fayl o'qish, filtrlash, guruhlash, birlashtrish va tezkor malumot tahlili

# Series ==> 1D(1 ta ustun)
import pandas as pd
import numpy as np

costs = pd.Series([620, 780, 1150, 380], name="narx")
print(costs)
```

```python
# DataFrame ==> butun jadval
homes = pd.DataFrame({
    "area":  [65, 90, 120, 45, 78],
    "rooms": [2, 3, 4, 1, 3],
    "location":   ["Yunusobod", "Chilonzor", "Mirzo Ulug'bek", "Sergeli", "Yakkasaroy"],
    "age":    [12, 5, 2, 25, 8],
    "cost":    [620, 780, 1150, 380, 810],
})

print(homes)
```

```python
print(homes.head(3))       # birinchi 3 qator
print(homes.tail(3))       # oxirgi 3 qator
print(homes.shape)         # (qatorlar, ustunlar)
print(homes.columns)       # ustun nomlari
print(homes.dtypes)
homes.info()
```

```python
print(homes.describe())
```

```python
homes["cost"]        # Series - bir o'lchovli
homes[["cost"]]      # DataFrame - ikki o'lchovli
```

```python
print(homes.iloc[0])          # pozitsiya bo'yicha
print(homes.iloc[1:3])        # 1 va 2-qator
print(homes.iloc[0, 2])       # birinchi qator, uchinchi ustun

print(homes.loc[0])           # indeks bo'yicha
print(homes.loc[0, "cost"])   # indeks va ustun nomi
print(homes.loc[0:2, ["area", "cost"]])
```

```python
expensive = homes[homes["cost"] > 700]
print(expensive)
```

```python
# Guruhlash
trade = pd.DataFrame({
    "location":  ["Chilonzor", "Chilonzor", "Sergeli", "Sergeli", "Yunusobod"],
    "rooms": [2, 3, 2, 1, 3],
    "cost":   [700, 780, 450, 380, 900],
})

print(trade.groupby("location")["cost"].mean())
```
